# 08 · Generalisation
Two tests of whether the models learned attack behaviour rather than artefacts of one capture:

1. **Cross-dataset:** every trained model evaluated, without retraining, on corrected CSE-CIC-IDS2018 (same fixed CICFlowMeter → identical features). Binary only, because the attack taxonomies differ.
2. **Leave-one-attack-out:** a binary detector trained without one attack family is tested on that family. This directly tests the 'detects attacks it has never seen' claim from the abstract.
